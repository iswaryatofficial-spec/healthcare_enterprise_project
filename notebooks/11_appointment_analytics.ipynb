```python
# ========================================
# 11_appointment_analytics
# ========================================

from pyspark.sql.functions import *

# Read Appointments Source Data
appointments_df = spark.read.format("csv") \
    .option("header", True) \
    .option("inferSchema", True) \
    .load(f"{source_path}/appointments")

# Generate Appointment KPIs
appointment_kpi_df = appointments_df.groupBy(
    "appointment_status"
).agg(
    count("appointment_id").alias("appointment_count")
)

# Write to Gold Layer
appointment_kpi_df.write \
    .format("delta") \
    .mode("overwrite") \
    .save(f"{gold_path}/appointment_kpis")

print("Appointment KPI Completed")
```

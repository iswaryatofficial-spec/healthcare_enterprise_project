```python
# ========================================
# 10_gold_pharma_analytics
# ========================================

from pyspark.sql.functions import *

# Read Pharmacy Source Data
pharma_df = spark.read.format("csv") \
    .option("header", True) \
    .option("inferSchema", True) \
    .load(f"{source_path}/pharmacy")

# Aggregate Medicine Sales
pharma_analytics_df = pharma_df.groupBy(
    "medicine_id"
).agg(
    sum("total_amount").alias("total_sales")
)

# Write to Gold Layer
pharma_analytics_df.write \
    .format("delta") \
    .mode("overwrite") \
    .save(f"{gold_path}/pharma_analytics")

print("Pharma Analytics Completed")
```

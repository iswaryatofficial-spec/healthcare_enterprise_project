```python
# ========================================
# 12_bed_management_analytics
# ========================================

from pyspark.sql.functions import *

# Read Bed Management Source Data
bed_df = spark.read.format("csv") \
    .option("header", True) \
    .option("inferSchema", True) \
    .load(f"{source_path}/bed_management")

# Generate Bed Occupancy Analytics
bed_analytics_df = bed_df.groupBy(
    "occupancy_status"
).agg(
    count("bed_id").alias("bed_count")
)

# Write to Gold Layer
bed_analytics_df.write \
    .format("delta") \
    .mode("overwrite") \
    .save(f"{gold_path}/bed_analytics")

print("Bed Analytics Completed")
```

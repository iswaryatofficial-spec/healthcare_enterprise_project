```python
# ========================================
# 08_gold_visit_analytics
# ========================================
 
from pyspark.sql.functions import *
 
# Read EMR Data
emr_df = spark.read.format("csv") \
    .option("header", True) \
    .option("inferSchema", True) \
    .load(f"{source_path}/emr")
 
# Create Visit Analytics
visit_analytics_df = emr_df.groupBy(
    "doctor_id",
    "procedure_code"
).agg(
    count("emr_id").alias("total_visits"),
    countDistinct("patient_id").alias("unique_patients")
)
 
# Write Gold Layer Table
visit_analytics_df.write \
    .format("delta") \
    .mode("overwrite") \
    .save(f"{gold_path}/patient_visit_analytics")
 
print("Visit Analytics Completed")
```

```python
# ========================================
# 18_governance_security
# ========================================

from pyspark.sql.functions import *

patients_df = (
    spark.read
        .format("delta")
        .load(f"{silver_path}/patients_clean")
)

secure_reporting_df = (
    patients_df.select(
        sha2(
            col("patient_id").cast("string"),
            256
        ).alias("masked_patient_id"),
        "city",
        "gender"
    )
)

(
    secure_reporting_df.write
        .format("delta")
        .mode("overwrite")
        .save(f"{gold_path}/secure_reporting")
)

print("Secure Reporting Layer Completed")
```
